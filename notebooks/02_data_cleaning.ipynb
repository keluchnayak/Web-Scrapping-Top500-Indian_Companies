# Step 2: Data Cleaning
In this notebook, we will clean the raw data scraped from AmbitionBox.


```
import pandas as pd
import numpy as np
import os

# Load raw dataset
raw_path = '../data/ambitionbox_top_500_raw.csv'
df = pd.read_csv(raw_path)

print(f"Raw data shape: {df.shape}")
df.head()
```

### 1. Check for missing values and duplicates


```
# Drop duplicate companies if any were scraped
df = df.drop_duplicates(subset=['company_url'])
print(f"Shape after dropping duplicates: {df.shape}")

# Check missing values
print("\nMissing values per column:")
print(df.isnull().sum())
```

### 2. Data Type Verification
Thanks to parsing the Next.js JSON directly, our numeric fields (ratings, reviews) are already mostly clean floats and integers instead of strings like '1.2k'! Let's just ensure they are treated correctly.


```
# Verify data types
df.info()
```

### 3. Cleaning Text Fields
Let's clean up the `headquarters` and `employees` columns.


```
# Clean Headquarters (often formatted as 'City, State, Country')
# Let's extract just the city for easier analysis
def extract_city(hq_str):
    if pd.isna(hq_str):
        return np.nan
    return str(hq_str).split(',')[0].strip()

df['hq_city'] = df['headquarters'].apply(extract_city)

# Clean Employees field if needed
df['employees_clean'] = df['employees'].str.replace('Employees', '').str.strip()

df[['headquarters', 'hq_city', 'employees', 'employees_clean']].head()
```

### 4. Save Cleaned Dataset


```
# Save cleaned data to CSV
cleaned_csv_path = '../data/ambitionbox_top_500_cleaned.csv'
df.to_csv(cleaned_csv_path, index=False)
print(f"Cleaned data saved to {cleaned_csv_path}")

# Also try saving to Excel
try:
    cleaned_excel_path = '../data/ambitionbox_top_500_cleaned.xlsx'
    df.to_excel(cleaned_excel_path, index=False)
    print(f"Cleaned data saved to {cleaned_excel_path}")
except Exception as e:
    print(f"Could not save Excel file (missing openpyxl?): {e}")
```
